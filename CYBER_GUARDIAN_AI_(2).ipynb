```python
# ==========================================
# CYBER GUARDIAN AI
# EMAIL PHISHING DETECTOR
# ==========================================

import re
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Sample training data
emails = [
    "Congratulations you won a prize click the link to claim",
    "Your account has been suspended verify your password immediately",
    "Urgent action required click here to confirm your bank account",
    "You have won a free gift provide your details to receive it",
    "Your payment failed login immediately to avoid account closure",
    "Security alert verify your account using the attached link",
    "Claim your reward now by clicking this link",
    "Your bank account needs verification urgently",
    "You are selected for a cash prize submit your information",
    "Reset your password immediately using this link",

    "Meeting is scheduled for tomorrow at 10 AM",
    "Please find the assignment attached",
    "Your order has been shipped successfully",
    "The project presentation is scheduled for Friday",
    "Please submit the report before Monday",
    "Thank you for attending today's meeting",
    "Your application has been received",
    "The class will begin at 9 AM tomorrow",
    "Please review the document and provide feedback",
    "Your library book is due next week"
]

labels = [
    "phishing", "phishing", "phishing", "phishing", "phishing",
    "phishing", "phishing", "phishing", "phishing", "phishing",

    "safe", "safe", "safe", "safe", "safe",
    "safe", "safe", "safe", "safe", "safe"
]

# Create AI model
email_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2)
    )),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train model
email_model.fit(emails, labels)

print("===================================")
print("🛡️ CYBER GUARDIAN AI")
print("===================================")
print("✅ Email phishing model trained!")
print("📧 Total training emails:", len(emails))
print("🔴 Phishing samples:", labels.count("phishing"))
print("🟢 Safe samples:", labels.count("safe"))
```

```python
#test email
```

```python
test_email = """
URGENT! Your bank account has been suspended.
Click the link immediately to verify your password.
"""

prediction = email_model.predict([test_email])[0]
probability = email_model.predict_proba([test_email]).max()

print("📧 EMAIL ANALYSIS")
print("----------------------------")
print("Prediction :", prediction.upper())
print("Confidence :", round(probability * 100, 2), "%")
```

```python
#add the email risk analyser
```

```python
def analyze_email(email):

    if not email.strip():
        return "⚠️ Please enter an email."

    prediction = email_model.predict([email])[0]

    suspicious_words = [
        "urgent",
        "verify",
        "password",
        "click",
        "winner",
        "prize",
        "bank",
        "suspended",
        "account",
        "claim",
        "payment",
        "security alert",
        "immediately",
        "confirm"
    ]

    found_words = []

    for word in suspicious_words:
        if word.lower() in email.lower():
            found_words.append(word)

    # Calculate risk
    score = 0

    if prediction == "phishing":
        score += 60

    score += min(len(found_words) * 5, 30)

    if "http://" in email.lower():
        score += 10

    score = min(score, 100)

    # Threat level
    if score >= 75:
        level = "🔴 HIGH RISK"
    elif score >= 45:
        level = "🟠 MEDIUM RISK"
    else:
        level = "🟢 LOW RISK"

    report = f"""
========================================
🛡️ CYBER GUARDIAN AI
EMAIL SECURITY REPORT
========================================

📧 AI Classification : {prediction.upper()}

⚠️ Threat Level     : {level}

📊 Risk Score       : {score}/100

🔍 Suspicious Words :
{", ".join(found_words) if found_words else "None detected"}

🛡️ Recommendation:
"""

    if score >= 75:
        report += """
DO NOT click links or provide personal
information. Verify the sender through
an official channel.
"""
    elif score >= 45:
        report += """
Be careful with links and attachments.
Verify the sender before responding.
"""
    else:
        report += """
No major warning signs were detected.
Continue normal security practices.
"""

    return report
```

```python
# ==========================================
# CYBER GUARDIAN AI
# URL RISK CHECKER
# ==========================================

from urllib.parse import urlparse

def analyze_url(url):

    if not url.strip():
        return "⚠️ Please enter a URL."

    original_url = url.strip()

    # Add protocol if missing
    if not re.match(r"^https?://", original_url, re.IGNORECASE):
        check_url = "http://" + original_url
    else:
        check_url = original_url

    parsed = urlparse(check_url)
    hostname = parsed.hostname or ""

    score = 0
    warnings = []

    # 1. HTTP instead of HTTPS
    if parsed.scheme.lower() == "http":
        score += 20
        warnings.append("Website uses HTTP instead of HTTPS")

    # 2. IP address instead of domain
    if re.match(r"^\d+\.\d+\.\d+\.\d+$", hostname):
        score += 30
        warnings.append("URL uses an IP address")

    # 3. Very long URL
    if len(original_url) > 75:
        score += 15
        warnings.append("URL is unusually long")

    # 4. @ symbol
    if "@" in original_url:
        score += 20
        warnings.append("URL contains an @ symbol")

    # 5. Too many subdomains
    if hostname.count(".") >= 3:
        score += 15
        warnings.append("URL contains many subdomains")

    # 6. Suspicious words
    suspicious_terms = [
        "login",
        "verify",
        "account",
        "secure",
        "update",
        "password",
        "free",
        "claim",
        "bonus",
        "bank"
    ]

    found_terms = []

    for term in suspicious_terms:
        if term in original_url.lower():
            found_terms.append(term)

    if found_terms:
        score += min(len(found_terms) * 5, 20)
        warnings.append(
            "Suspicious terms found: " +
            ", ".join(found_terms)
        )

    score = min(score, 100)

    # Threat level
    if score >= 70:
        level = "🔴 HIGH RISK"
    elif score >= 40:
        level = "🟠 MEDIUM RISK"
    else:
        level = "🟢 LOW RISK"

    report = f"""
========================================
🛡️ CYBER GUARDIAN AI
URL SECURITY REPORT
========================================

🔗 URL:
{original_url}

⚠️ Threat Level : {level}

📊 Risk Score   : {score}/100

🔍 FINDINGS
"""

    if warnings:
        for warning in warnings:
            report += f"\n• {warning}"
    else:
        report += "\n• No major suspicious characteristics detected."

    report += """

🛡️ RECOMMENDATION
"""

    if score >= 70:
        report += """
Avoid visiting this URL or entering
personal information.
"""
    elif score >= 40:
        report += """
Proceed carefully and verify the website
independently before entering information.
"""
    else:
        report += """
No major URL warning signs detected.
"""

    return report
```

```python
# ==========================================
# CYBER GUARDIAN AI
# PASSWORD STRENGTH CHECKER
# ==========================================

def analyze_password(password):

    if not password:
        return "⚠️ Please enter a password."

    score = 0
    suggestions = []

    # Length
    if len(password) >= 8:
        score += 25
    else:
        suggestions.append("Use at least 8 characters")

    if len(password) >= 12:
        score += 15

    # Uppercase
    if re.search(r"[A-Z]", password):
        score += 15
    else:
        suggestions.append("Add uppercase letters")

    # Lowercase
    if re.search(r"[a-z]", password):
        score += 10
    else:
        suggestions.append("Add lowercase letters")

    # Numbers
    if re.search(r"\d", password):
        score += 15
    else:
        suggestions.append("Add numbers")

    # Special characters
    if re.search(r"[^A-Za-z0-9]", password):
        score += 20
    else:
        suggestions.append("Add special characters")

    score = min(score, 100)

    # Strength level
    if score >= 80:
        level = "🟢 STRONG"
    elif score >= 50:
        level = "🟠 MEDIUM"
    else:
        level = "🔴 WEAK"

    report = f"""
========================================
🛡️ CYBER GUARDIAN AI
PASSWORD SECURITY REPORT
========================================

🔐 Strength Level : {level}

📊 Security Score : {score}/100

💡 SECURITY SUGGESTIONS
"""

    if suggestions:
        for suggestion in suggestions:
            report += f"\n• {suggestion}"
    else:
        report += "\n• Good password characteristics detected."

    report += """

🛡️ SECURITY TIP

Use a unique password for each important
account and avoid using personal information.
"""

    return report
```

```python
# ============================================================
# 🛡️ CYBER GUARDIAN AI
# AI-POWERED CYBERSECURITY ASSISTANT
# ============================================================

import gradio as gr

# ------------------------------------------------------------
# Dashboard
# ------------------------------------------------------------

custom_css = """
body {
    background: #0b1020;
}

.gradio-container {
    max-width: 1100px !important;
    margin: auto !important;
}

h1 {
    text-align: center;
    font-size: 38px !important;
}

.dashboard-card {
    padding: 20px;
    border-radius: 15px;
}
"""

with gr.Blocks(
    title="Cyber Guardian AI",
    css=custom_css,
    theme=gr.themes.Soft()
) as app:

    # ========================================================
    # HEADER
    # ========================================================

    gr.Markdown("""
# 🛡️ CYBER GUARDIAN AI

### AI-Powered Cybersecurity Assistant

**Detect • Analyze • Protect**

Cyber Guardian AI helps identify common cybersecurity
risks in emails, URLs and passwords.
""")

    gr.Markdown("""
---

### 🔍 Available Security Tools

📧 **Email Scanner** &nbsp;&nbsp;
🔗 **URL Scanner** &nbsp;&nbsp;
🔐 **Password Checker**

---
""")

    # ========================================================
    # TABS
    # ========================================================

    with gr.Tabs():

        # ====================================================
        # EMAIL SCANNER
        # ====================================================

        with gr.Tab("📧 Email Scanner"):

            gr.Markdown("""
## 📧 Phishing Email Detector

Paste an email message below. Cyber Guardian AI will
analyze it for common phishing indicators.
""")

            email_input = gr.Textbox(
                label="Email Content",
                placeholder="Paste the email message here...",
                lines=10
            )

            email_button = gr.Button(
                "🔍 Analyze Email",
                variant="primary"
            )

            email_output = gr.Markdown(
                label="Security Report"
            )

            email_button.click(
                analyze_email,
                inputs=email_input,
                outputs=email_output
            )

        # ====================================================
        # URL SCANNER
        # ====================================================

        with gr.Tab("🔗 URL Scanner"):

            gr.Markdown("""
## 🔗 Suspicious URL Detector

Enter a website URL to check for common suspicious
characteristics.
""")

            url_input = gr.Textbox(
                label="Website URL",
                placeholder="https://example.com"
            )

            url_button = gr.Button(
                "🔍 Analyze URL",
                variant="primary"
            )

            url_output = gr.Markdown(
                label="Security Report"
            )

            url_button.click(
                analyze_url,
                inputs=url_input,
                outputs=url_output
            )

        # ====================================================
        # PASSWORD CHECKER
        # ====================================================

        with gr.Tab("🔐 Password Checker"):

            gr.Markdown("""
## 🔐 Password Strength Analyzer

Check whether a password has common characteristics
of a stronger password.
""")

            password_input = gr.Textbox(
                label="Password",
                placeholder="Enter a password",
                type="password"
            )

            password_button = gr.Button(
                "🔍 Check Password",
                variant="primary"
            )

            password_output = gr.Markdown(
                label="Security Report"
            )

            password_button.click(
                analyze_password,
                inputs=password_input,
                outputs=password_output
            )

    # ========================================================
    # FOOTER
    # ========================================================

    gr.Markdown("""
---

## 🛡️ Cyber Guardian AI

**Modules:**
📧 Phishing Detection • 🔗 URL Analysis • 🔐 Password Strength

### ⚠️ Disclaimer

This project is an educational cybersecurity prototype.
It uses simplified detection techniques and should not
be considered a replacement for professional security
software.

**Stay Alert • Think Before You Click • Stay Secure 🔐**
""")

# ============================================================
# LAUNCH
# ============================================================

app.launch(
    share=True,
    inline=True
)
```
